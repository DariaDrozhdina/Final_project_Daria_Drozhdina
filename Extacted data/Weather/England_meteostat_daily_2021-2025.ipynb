## Monthly weather data for England

```python
import requests
import pandas as pd
import time
from dotenv import dotenv_values

config = dotenv_values()
api_key = config['x_rapidapikey'] 

API_KEY = api_key
HEADERS = {
    "X-RapidAPI-Key": API_KEY,
    "X-RapidAPI-Host": "meteostat.p.rapidapi.com"
}

STATIONS = {
    "03245": "Newcastle upon Tyne",
    "03262": "Sunderland",
    "03263": "Middlesbrough",
    "03347": "Leeds",
    "03334": "Manchester",
    "03266": "York",
    "03354": "Nottingham",
    "03462": "Peterborough",
    "03534": "Birmingham",
    "03827": "Plymouth",
    "03839": "Exeter",
    "03726": "Bristol",
    "03779": "London",
    "03876": "Brighton",
    "03801": "Canterbury"
}

all_records = []

for station_id, station_name in STATIONS.items():
    response = requests.get(
        "https://meteostat.p.rapidapi.com/stations/monthly",
        headers=HEADERS,
        params={"station": station_id, "start": "2021-01-01", "end": "2025-12-31"}
    )
    data = response.json().get("data", [])

    for row in data:
        row["station_id"]   = station_id
        row["station_name"] = station_name
        all_records.append(row)

    print(f"✅ {station_name}: {len(data)} months fetched")
    time.sleep(0.3)

df = pd.DataFrame(all_records)

print(df)
df.to_csv("England_weather.csv", index=False)
```
