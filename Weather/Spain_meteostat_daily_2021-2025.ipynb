## Monthly weather data for Spain

```python
import requests
import pandas as pd
import time
from dotenv import dotenv_values

config = dotenv_values()
api_key = config['x_rapidapikey'] 

API_KEY = api_key
HEADERS = {
    "X-RapidAPI-Key": API_KEY,
    "X-RapidAPI-Host": "meteostat.p.rapidapi.com"
}

STATIONS = {
    "08025": "Bilbao",
    "08042": "Santiago de Compostela",
    "08284": "Valencia",
    "08181": "Barcelona",
    "08391": "Seville",
    "08482": "Malaga",
    "08221": "Madrid",
    "08160": "Zaragoza"
}

all_records = []

for station_id, station_name in STATIONS.items():
    response = requests.get(
        "https://meteostat.p.rapidapi.com/stations/monthly",
        headers=HEADERS,
        params={"station": station_id, "start": "2021-01-01", "end": "2025-12-31"}
    )
    data = response.json().get("data", [])

    for row in data:
        row["station_id"]   = station_id
        row["station_name"] = station_name
        all_records.append(row)

    print(f"✅ {station_name}: {len(data)} months fetched")
    time.sleep(0.3)

df = pd.DataFrame(all_records)

print(df)
df.to_csv("spain_weather.csv", index=False)
```
